```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Load data
train = pd.read_csv('../data/train.csv')
test = pd.read_csv('../data/test.csv')
sample_sub = pd.read_csv('../data/sample_submission.csv')

print(f'Train shape: {train.shape}')
print(f'Test shape: {test.shape}')
print(f'Sample submission shape: {sample_sub.shape}')
```

```python
# Check missing values
missing_train = train.isnull().sum().sort_values(ascending=False)
missing_test = test.isnull().sum().sort_values(ascending=False)

print('=== Missing values in Train ===')
print(missing_train[missing_train > 0])

print('\n=== Missing values in Test ===')
print(missing_test[missing_test > 0])
```

```python
# Target distribution
print('Target distribution:')
print(train['addicted_label'].value_counts())
print(f"\nAddiction rate: {train['addicted_label'].mean():.4f}")
```

```python
# Categorical features analysis
cat_cols = ['gender', 'stress_level', 'academic_work_impact']
for col in cat_cols:
    print(f'=== {col} ===')
    print(train[col].value_counts(dropna=False))
    print()
```

```python
# Correlation with target
num_cols = train.select_dtypes(include=[np.number]).columns.tolist()
num_cols.remove('id')
num_cols.remove('addicted_label')

correlations = train[num_cols + ['addicted_label']].corr()['addicted_label'].sort_values(ascending=False)
print(correlations)
```
